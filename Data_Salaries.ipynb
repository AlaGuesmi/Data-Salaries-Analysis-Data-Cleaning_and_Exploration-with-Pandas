## Filling missing values

### Methods for filling NaaN (Impute):
- Mean/Median for numerical columns
- Mode for categorical columns
- Least occured value ≠ Mode for categorical columns
- bfill(), ffill() from pandas;  bfill : backward-filling;  ffill : forward-filling (classic methods)

- sklearn.impute: IterativeImputer, KNNImputer, MissingIndicator, SimpleImputer (more advanced methods)
- feature engine: Random Imputation, CategoricalImputer

- Listewise/Pairwise deletion (more complex in particual cases )

```python
mv 
#mv is just a snapshot of the missing values at that moment. It doesn't dynamically update after you modify the DataFrame.
```

```python
df['job_title']=df['job_title'].fillna(df['job_title'].mode()[0])
#Rks: mode() displays index and mode, so we write mode()[0] to get the mode only
#i can use inplace=True, in that case no need to create a new variable and assign the function to it. We apply the fucntion directly and it changes the orginal data
#what we did: Create a new Series and assigns it back to df['salary'].
```

```python
df['job_title'].mode() 
```

```python
type(df['job_title'].mode()) #the type of mode function is series like array or dataframe so to slice we need []
```

```python
#to chack if there is a change
df.isnull().sum()
```

```python
df['salary'].mean()
```

```python
df['salary']=df['salary'].fillna(df['salary'].mean())
```

```python
df['salary_in_usd']=df['salary_in_usd'].fillna(df['salary_in_usd'].mean())
```

```python
df.isnull().sum()
```

```python
df['salary_currency'].value_counts() #occurences, type is series (it displays name and dtype)
#value_counts returns series. Then USD, etc are index and in front of them are the values
```

```python
least_occurred_value=df['salary_currency'].value_counts().idxmin() #idxmin and idxmax show least and most occured values
least_occurred_value
```

```python
df['salary_currency']=df['salary_currency'].fillna(least_occurred_value) #we can fill missing values with 0 (in numerical cols)
```

```python
df.isnull().sum() 
```

```python
df['experience_level']=df['experience_level'].fillna('N/A')
```

```python
df.isnull().sum()
```

```python
for col in ['employment_type','company_location','company_size']:
    df[col]=df[col].fillna(df[col].bfill())
```

```python
sum(df.isnull().sum())
```
